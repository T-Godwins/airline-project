## Data Preprocessing

```python
df.dropna(inplace=True)
df.isnull().sum()
```

```python
# Encode Categorical Data
df_train = df.copy()
df_train['label'] = df_train['satisfaction'].apply(lambda x:1 if x == 'satisfied' else 0)

encode_cols = ["Gender", "Customer Type", "Type of Travel", "Class"]
df_encoded = pd.get_dummies(df[encode_cols], columns=encode_cols)
df_encoded = df_encoded.astype(int)

display(df_encoded.head())
```

```python
# Drop unnecesaary columns
dropped_cols = ['Gender', 'Customer Type', 'Type of Travel', 'Class', 'satisfaction']
df_train = df_train.drop(dropped_cols, axis=1)
```

### Model Training

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Create train dataset
X_train = y_train = df_train[df_train['type'] == 'train']
X_train = X_train.drop(['type', 'label'], axis=1)
y_train = y_train['label']

# Create test dataset
X_test = y_test = df_train[df_train['type'] == 'test']
X_test = X_test.drop(['type', 'label'], axis=1)
y_test = y_test['label']

# Create model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Evaluate model
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
```

```python
print(f"Accuracy: {accuracy}")
print(classification_report(y_test, y_pred))
```

### Visualization Model ###

```python
from sklearn.tree import plot_tree
from sklearn.inspection import permutation_importance

def plot_feature_importance(model, importance_type, feature_names, n_features=10, figsize=(10,6)):
    # Get feature importances
    if importance_type == 'feature':
        importances = model.feature_importances_
    elif importance_type == 'permutation':
        importances = permutation_importance(model, X_test, y_test, n_repeats=10, random_state=42).importances_mean
    
    # Sort features by importance
    indices = np.argsort(importances)[::-1]
    
    # Plot top n_features
    plt.figure(figsize=figsize)
    plt.title('Top Feature Importances in Sentiment Analysis')
    
    # Plot only top n_features
    n = min(n_features, len(importances))
    plt.bar(range(n), 
            importances[indices][:n],
            align='center')
    
    # Use actual words as labels
    plt.xticks(range(n), 
               [feature_names[i] for i in indices[:n]], 
               rotation=45,
               ha='right')
    
    plt.xlabel('Features')
    plt.ylabel('Importance Score')
    plt.tight_layout()
    plt.show()


def plot_decision_tree(model, max_depth=3, figsize=(20,10)):
    plt.figure(figsize=figsize)
    # Get a single tree from the forest
    tree = model.estimators_[0]
    
    # Create feature names for all dimensions
    feature_names = [f'dim_{i}' for i in range(model.n_features_in_)]
    
    plot_tree(tree, 
             feature_names=feature_names,
             max_depth=max_depth,
             filled=True,
             rounded=True)
    plt.show()
```

```python
import warnings
warnings.filterwarnings("ignore")

# Empty dictionaries to store importances and feature names
all_importances_feature = {}
all_importances_permutation = {}
all_feature_names = set()
```

```python
# Define original features that will be one-hot encoded
original_features = ["Gender", "Customer Type", "Type of Travel", "Class"]
all_importances = {}
all_feature_names = set()

for tier in df['Class'].unique():
    print(f"\nProcessing {tier}")
    
    # Get data for this tier
    df_tier = df[df['Class'] == tier]
    
    # Create encoded features
    df_encoded = pd.get_dummies(df_tier[original_features], columns=original_features)
    df_encoded = df_encoded.astype(int)
    
    # Add encoded columns back
    df_processed = df_tier.drop(original_features, axis=1)
    df_processed = pd.concat([df_processed, df_encoded], axis=1)
    
    # Create label
    df_processed['label'] = df_processed['satisfaction'].apply(lambda x: 1 if x == 'satisfied' else 0)
    
    # Split data
    train_mask = df_processed['type'] == 'train'
    df_train = df_processed[train_mask]
    
    # Prepare features and target
    dropped_cols = ['type', 'label', 'satisfaction']
    X_tier = df_train.drop(dropped_cols, axis=1)
    y_tier = df_train['label']
    # print(X_tier.columns)
    
    # Train model
    model = RandomForestClassifier(n_estimators=100, random_state=42)
    model.fit(X_tier, y_tier)
    
    # Calculate permutation importance
    perm_importance = permutation_importance(model, X_tier, y_tier, n_repeats=10, random_state=42)
    
    # Aggregate importance scores for encoded features
    aggregated_feature_importance = {}
    aggregated_perm_importance = {}
    
    # Process feature importances
    for col, imp_feature, imp_perm in zip(X_tier.columns, model.feature_importances_, perm_importance.importances_mean):
        # Check if column is from one-hot encoding
        matched = False
        for feature in original_features:
            if col.startswith(feature + "_"):
                aggregated_feature_importance[feature] = aggregated_feature_importance.get(feature, 0) + imp_feature
                aggregated_perm_importance[feature] = aggregated_perm_importance.get(feature, 0) + imp_perm
                matched = True
                break
        # If not encoded, use original column name
        if not matched:
            aggregated_feature_importance[col] = imp_feature
            aggregated_perm_importance[col] = imp_perm
    
    # Store aggregated importances
    all_importances[tier] = {
        'feature_importances': list(aggregated_feature_importance.values()),
        'permutation_importances': list(aggregated_perm_importance.values()),
        'features': list(aggregated_feature_importance.keys())
    }
    all_feature_names.update(aggregated_feature_importance.keys())
```

```python
for i in all_importances:
    print(all_importances[i])
# display(all_importances)  
```

```python
all_feature_names = sorted(list(all_feature_names))

# Calculate average importance across tiers to determine sorting
avg_feature_importance = np.zeros(len(all_feature_names))
avg_permutation_importance = np.zeros(len(all_feature_names))

for data in all_importances.values():
    for i, feature in enumerate(data['features']):
        feature_idx = all_feature_names.index(feature)
        avg_feature_importance[feature_idx] += data['feature_importances'][i]
        avg_permutation_importance[feature_idx] += data['permutation_importances'][i]

avg_feature_importance /= len(all_importances)
avg_permutation_importance /= len(all_importances)

# Sort features by average importance
feature_sort_idx = np.argsort(avg_feature_importance)[::-1]
perm_sort_idx = np.argsort(avg_permutation_importance)[::-1]

sorted_feature_names_fi = [all_feature_names[i] for i in feature_sort_idx]
sorted_feature_names_pi = [all_feature_names[i] for i in perm_sort_idx]

# Create x positions for bars
x = np.arange(len(all_feature_names))

# Create figures for feature importance and permutation importance
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(15, 16))
bar_width = 0.15
opacity = 0.8

# Plot feature importance bars for each tier
for idx, (tier, data) in enumerate(all_importances.items()):
    # Create importance array matching sorted features
    importances = np.zeros(len(all_feature_names))
    for i, feature in enumerate(data['features']):
        feature_idx = sorted_feature_names_fi.index(feature)
        importances[feature_idx] = data['feature_importances'][i]
    
    ax1.bar(x + idx * bar_width, 
            importances,
            bar_width,
            alpha=opacity,
            label=tier)

# Plot permutation importance bars for each tier
for idx, (tier, data) in enumerate(all_importances.items()):
    # Create importance array matching sorted features
    importances = np.zeros(len(all_feature_names))
    for i, feature in enumerate(data['features']):
        feature_idx = sorted_feature_names_pi.index(feature)
        importances[feature_idx] = data['permutation_importances'][i]
    
    ax2.bar(x + idx * bar_width, 
            importances,
            bar_width,
            alpha=opacity,
            label=tier)

# Customize feature importance plot
ax1.set_xlabel('Features')
ax1.set_ylabel('Feature Importance Score')
ax1.set_title('Feature Importance by Class (Sorted by Average Importance)')
ax1.set_xticks(x + bar_width * (len(all_importances) - 1) / 2)
ax1.set_xticklabels(sorted_feature_names_fi, rotation=45, ha='right')
ax1.legend()

# Customize permutation importance plot  
ax2.set_xlabel('Features')
ax2.set_ylabel('Permutation Importance Score')
ax2.set_title('Permutation Importance by Class (Sorted by Average Importance)')
ax2.set_xticks(x + bar_width * (len(all_importances) - 1) / 2)
ax2.set_xticklabels(sorted_feature_names_pi, rotation=45, ha='right')
ax2.legend()

plt.tight_layout()
plt.show()
```
